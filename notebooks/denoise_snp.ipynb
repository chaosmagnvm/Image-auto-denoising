# salt and pepper noise

random pixels replaced with extreme values: black (value=0) or white (value=255)

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
import dss.filters as f
import dss.detector as d

pic = 28
img = cv2.imread(f"../noise_gauss/original_png/00{pic}.png")
img = f.snp_noise(img, p= 0.01)

fig, axes = plt.subplots(1, 3, figsize = (16, 10))

denoised = f.denoiser_snp(img, d.snp_noise_detector(img))
rgb_denoised = cv2.cvtColor(denoised, cv2.COLOR_BGR2RGB)

rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

axes[0].imshow(rgb)
axes[0].set_title("Noisy (p=0.01)")
axes[0].axis("off")

axes[1].imshow(rgb_denoised)
axes[1].set_title("Restored")
axes[1].axis("off")

src = cv2.imread(f"../noise_gauss/original_png/00{pic}.png")
rgb_clean = cv2.cvtColor(src, cv2.COLOR_BGR2RGB)
axes[2].imshow(rgb_clean)
axes[2].set_title("Source (clean)")
axes[2].axis("off");
```
