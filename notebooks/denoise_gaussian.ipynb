# Gaussian noise auto-restore

Gaussian noise has a normal distribution. To restore an image filter requires it's std (standart deviation). 

```python
import numpy as np
import cv2
import matplotlib.pyplot as plt
import dss.filters as f
import dss.detector as d

pic = 43
img = cv2.imread(f"../noise_gauss/noisy35/00{pic}.png")

denoised = img.copy()
denoised = f.denoiser_gaussian(img, d.gaussian_noise_detector(img))

fig, axes = plt.subplots(1,3, figsize=(15,10))

rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
rgb_clear = cv2.cvtColor(denoised, cv2.COLOR_BGR2RGB)

axes[0].imshow(rgb)
axes[0].set_title("Noised(std=35)")
axes[0].axis("off")

axes[1].imshow(rgb_clear)
axes[1].set_title("Restored")
axes[1].axis("off")

src = cv2.imread(f"../noise_gauss/original_png/00{pic}.png")
rgb_clean = cv2.cvtColor(src, cv2.COLOR_BGR2RGB)
axes[2].imshow(rgb_clean)
axes[2].set_title("Source (clean)")
axes[2].axis("off");
```
